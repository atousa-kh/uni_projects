```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.RandomState(0)

fig, ax = plt.subplots(figsize=(3, 3))

n_blue, n_smalll , n_small = 1000, 1000, 1000

X_blue_rect = np.c_[rng.uniform(-2, 0, n_blue),
                    rng.uniform(-1.2, 1.2, n_blue)]

X_orange_rect = np.c_[rng.uniform(0, 2, n_smalll),
                      rng.uniform(0, 1.2, n_smalll)]

X_green_rect = np.c_[rng.uniform(0, 2, n_small),
                     rng.uniform(-1.2, 0, n_small)]

clusters6 = [X_blue_rect, X_orange_rect, X_green_rect]
colors6 = ['C0', 'C1', 'C2']

for pts, col in zip(clusters6, colors6):
    ax.scatter(pts[:, 0], pts[:, 1], s=8, color=col)

ax.set_xticks([]); ax.set_yticks([])
ax.text(0.98, 0.02, ".01s", transform=ax.transAxes,
        ha='right', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```

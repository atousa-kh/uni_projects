```python
from sklearn.datasets import make_blobs  # import module to make blobs
import matplotlib.pyplot as plt
import numpy as np

# make blobs function
def make_blob():
    centers = np.array([[5, 5],   
                        [0, 0],   
                        [-5, -5]])
    stds=[0.5, 2.5, 1.0]
    x, y = make_blobs(n_samples=1500, centers=centers,
                      cluster_std=stds, random_state=42,
                      shuffle=True)
    return x, y


# generate data
x3 , y3 = make_blob()

# plot blobs
plt.figure(figsize=(5,5))
plt.scatter(x3[y3 == 0, 0], x3[y3 == 0, 1],
            color='blue', label='Right Top')
plt.scatter(x3[y3 == 1, 0], x3[y3 == 1, 1],
            color='orange', label='Middle')
plt.scatter(x3[y3 == 2, 0], x3[y3 == 2, 1],
            color='green', label='Left Bottom')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title("Custom Unequal Variance Clusters")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sqr_dataset():
    # number of samples for each region
    n_samples = [800, 400, 400]
    
    # left side of squere
    x1 = np.random.uniform(0, 0.5, n_samples[0])
    y1 = np.random.uniform(0, 1, n_samples[0])
    labels1 = np.zeros(n_samples[0], dtype=int)  # label 0 
    
    # upper half 
    x2 = np.random.uniform(0.5, 1.0, n_samples[1])
    y2 = np.random.uniform(0.5, 1.0, n_samples[1])
    labels2 = np.ones(n_samples[1], dtype=int)   # label 1
    
    # lower half
    x3 = np.random.uniform(0.5, 1.0, n_samples[2])
    y3 = np.random.uniform(0.0, 0.5, n_samples[2])
    labels3 = np.full(n_samples[2], 2, dtype=int) # label 2
    
    #  concatenate all points 
    X = np.concatenate([np.column_stack([x1, y1]),
                        np.column_stack([x2, y2]),
                        np.column_stack([x3, y3])], axis=0)
    y = np.concatenate([labels1, labels2, labels3], axis=0)
    return X, y
# generate data
x6, y6 = sqr_dataset()

#  plot squere
plt.figure(figsize=(5,5))
plt.scatter(x6[y6==0,0], x6[y6==0,1], color='blue', label='Left Half')
plt.scatter(x6[y6==1,0], x6[y6==1,1], color='orange', label='Top Right')
plt.scatter(x6[y6==2,0], x6[y6==2,1], color='green', label='Bottom Right')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Square Dataset Divided into Three Regions')
plt.show()
```

### Define Type, Operators, Function set, Terminal Set

```python
import math
from deap import gp, base, creator, tools

class PosFloat(float): # Positive float marker type for STGP.
    pass

def add(a, b): 
    return a + b

# def sub (a, b):
    return a - b

def mul(a, b): 
    return a * b

def neg(a):
    return -a

def safe_div(a, b):
    return a / b if abs(b) > 1e-12 else 1.0

# def safe_sqrt(x):
#     return math.sqrt(abs(x))

# def safe_log(x):
#     return math.log(1.0 + abs(x))

# def safe_abs(x):
#     return abs(x)

# def clamp_pos(x, eps=1e-12):
#     # insure to be positive
#     v = abs(x)
#     return v if v > eps else eps
    
# =======================================================================

# Primitive Set Typed (Function set type)
pset = gp.PrimitiveSetTyped(
    "DIST",
    [PosFloat, PosFloat, PosFloat, PosFloat],   
    PosFloat
)

pset.renameArguments(ARG0="xi")
pset.renameArguments(ARG1="yi")
pset.renameArguments(ARG2="xj")
pset.renameArguments(ARG3="yj")

# Functon set
pset.addPrimitive(add,      [PosFloat, PosFloat], PosFloat)
# pset.addPrimitive(sub,      [PosFloat, PosFloat], PosFloat)
pset.addPrimitive(mul,      [PosFloat, PosFloat], PosFloat)
pset.addPrimitive(neg,      [PosFloat],           PosFloat)
pset.addPrimitive(safe_div, [PosFloat, PosFloat], PosFloat)
# pset.addPrimitive(safe_sqrt,[PosFloat],           PosFloat)
# pset.addPrimitive(safe_log, [PosFloat],           PosFloat)
# pset.addPrimitive(safe_abs, [PosFloat],           PosFloat)
# pset.addPrimitive(clamp_pos,[PosFloat],           PosFloat)

# Terminal set (some constants)
pset.addTerminal(0.5, PosFloat)
pset.addTerminal(1.0, PosFloat)
pset.addTerminal(2.0, PosFloat)
pset.addTerminal(5.0, PosFloat)
pset.addTerminal(-1.0, PosFloat)
```

#### Distance Function

```python
def distance(distance_func, X):  
   
    n = X.shape[0] # rows --> number of data
    W = np.zeros((n, n), dtype=float)

    for i in range(n):
        xi, yi = X[i, 0], X[i, 1]
        
        for j in range(i + 1, n):
            xj, yj = X[j, 0], X[j, 1]
            
            dij = distance_func(PosFloat(xi), PosFloat(yi),
                                PosFloat(xj), PosFloat(yj))
            if not np.isfinite(dij):
                return None

            dij = float(dij)
            dij = max(dij, 0.0) # negative numbers are not allowed

            # Symmetrize distance matrix
            dji = distance_func(PosFloat(xj), PosFloat(yj),
                                PosFloat(xi),PosFloat(yi))
            if not np.isfinite(dji): 
                return None

            dji = max(float(dji), 0.0)
            dij = 0.5 * (dij + dji)

           
            W[i, j] = dij
            W[j, i] = dij

    return W
```

#### Spectral Clustering Function

```python
import numpy as np
from sklearn.cluster import SpectralClustering

def spectral_with_distance(distance_func, X, n_clusters, random_state=0):
    W = distance(distance_func, X)

    if W is None:
        return None

    mx = np.max(W)
    if mx <= 1e-12 or not np.isfinite(mx):
        return None

    W = W / mx
    np.fill_diagonal(W, 1.0)
    
    spectral = SpectralClustering(
        n_clusters=n_clusters,
        affinity="precomputed",
        assign_labels="kmeans",
        random_state=random_state
    )
    labels = spectral.fit_predict(W)
    return labels
```

#### Fisher function

```python
def fisher(X, labels, eps=1e-12):
   
    labels = np.asarray(labels)
    unique = np.unique(labels)
    if unique.size < 2:    # if num of clusters < 2
        return 0.0

    mu = X.mean(axis=0, keepdims=True) # mean of all data
    SB = 0.0  # between clusters
    SW = 0.0  # within clusters

    for c in unique:
        Xc = X[labels == c]
        if Xc.shape[0] < 2:
            continue
        muc = Xc.mean(axis=0, keepdims=True) # mean of the cluster
        # between
        SB += Xc.shape[0] * float(np.sum((muc - mu) ** 2))
        # within
        SW += float(np.sum((Xc - muc) ** 2))

    return float(SB / (SW + eps))
```

```python
X = x1
y_true = y1
k = 2
```

#### Infrastructure of deap library

```python
# creator (Create Individual & FitnessMax Classes)

if "FitnessMax" not in creator.__dict__:
    creator.create("FitnessMax", base.Fitness, weights=(1.0,))
if "Individual" not in creator.__dict__:
    creator.create("Individual", gp.PrimitiveTree, fitness=creator.FitnessMax)
    
# =============================================================================

# Define structures to make individual and population
toolbox = base.Toolbox()
toolbox.register("expr", gp.genHalfAndHalf, pset=pset, min_=1, max_=4)
toolbox.register("individual", tools.initIterate, creator.Individual, toolbox.expr)
toolbox.register("population", tools.initRepeat, list, toolbox.individual)
toolbox.register("compile", gp.compile, pset=pset)
```

#### knn sub_set

```python
import random
from sklearn.neighbors import NearestNeighbors

def knn_subset(X, k=2, seed=123):
    rng = np.random.default_rng(seed)

    # randomly select the first point
    start_idx = rng.integers(0, X.shape[0]) 
    
    nbrs = NearestNeighbors(n_neighbors=k)
    nbrs.fit(X)
    distances, indices = nbrs.kneighbors(X[start_idx].reshape(1, -1))

    return X[indices[0]]

X_eval = knn_subset(X)

# def subset(X, size=200, seed=123):
#     rng = np.random.default_rng(seed)
#     idx = rng.choice(X.shape[0], size=size, replace=False)
    
#     return X[idx]

# X_eval = subset(X, size=200, seed=123)
```

#### Fitness Function

```python
def fitness(individual, X_sub, n_clusters):
    distance_func = toolbox.compile(expr=individual)

    labels = spectral_with_distance(distance_func, X_sub, n_clusters=n_clusters)

    score = fisher(X_sub, labels)

    return (score,)
```

#### Define Evolutionary Operators

```python
import operator

# Evaluate structures 
toolbox.register("evaluate", fitness, X_sub=X_eval, n_clusters=k)

# =============================================================================

# Genetic operators (Selection, Crossover, Mutation)
toolbox.register("select", tools.selTournament, tournsize=4) # Selection
toolbox.register("mate", gp.cxOnePoint) # Crossover
toolbox.register("expr_mut", gp.genFull, min_=0, max_=2)
toolbox.register("mutate", gp.mutUniform, expr=toolbox.expr_mut, pset=pset) # Mutation

# ============================================================================

# Tree growth control
toolbox.decorate("mate", gp.staticLimit(key=operator.attrgetter("height"), max_value=10))
toolbox.decorate("mutate", gp.staticLimit(key=operator.attrgetter("height"), max_value=10))
```

#### Genetic Programming Block to Evolve Distance Measure

```python
import numpy as np
from deap import algorithms

# Genetic Programming Block
def evolve_dist_func(n_pop=30, n_gen=20, 
                     cxpb=0.5, mutpb=0.3, seed=0):
    random.seed(seed)
    np.random.seed(seed)

    pop = toolbox.population(n=n_pop) # generate a population

    # display avg, min, max fitness value for each generation
    stats = tools.Statistics(lambda ind: ind.fitness.values[0]) 
    stats.register("avg", np.mean)
    stats.register("min", np.min)
    stats.register("max", np.max)

    hof = tools.HallOfFame(1) # keep the best individual during all generations

    # Algorithm
    pop, log = algorithms.eaSimple(
        pop, toolbox,
        cxpb=cxpb, mutpb=mutpb,
        ngen=n_gen,
        stats=stats,
        halloffame=hof,
        verbose=True  # Display report during evolutionery
    )

    return pop, log, hof  # log = statistics report of generations

```

#### Run GP 

```python
pop, log, hof = evolve_dist_func(
    n_pop=30,
    n_gen=25,
    cxpb=0.5,
    mutpb=0.3,
    seed=0
)

best_ind = hof[0]
best_ind, best_ind.fitness.values
```

#### Apply Spectral Clustering with GP Distance Measure

```python
from sklearn.metrics import adjusted_rand_score

best_func = toolbox.compile(expr=best_ind)

best_labels = spectral_with_distance(best_func, X, n_clusters=k)

print("Best Individual:", best_ind)
print("Fisher (on full X):", fisher(X, best_labels))
print("ARI vs true labels:", adjusted_rand_score(y_true, best_labels))

plt.figure(figsize=(6,5))
plt.scatter(X[:,0], X[:,1], c=best_labels, s=10)
plt.title("Spectral Clustering with Evolved Distance")
plt.xlabel("x1"); plt.ylabel("x2")
plt.show()
```
